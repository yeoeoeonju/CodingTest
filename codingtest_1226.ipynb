```python
# 게임 개발

count = 0


n, m = map(int, input().split())  # 맵 길이 입력받기

d = [[0] * m for _ in range(n)]  # 초기 맵 생성

x, y, direction = map(int, input().split()) # 현재 위치 및 방향 입력받기

d[x][y] = 1 # 지금있는 곳 방문했으므로 1로 변경하기 

# -----------------------------------------------------------


array = []

for i in range(n):
    array.append(list(map(int, input().split()))) # 바다, 육지 맵 입력받기 

dx = [-1, 0, 1, 0]
dy = [0, 1, 0, -1] #북, 동, 남, 서 이동 시 변경 정의

# --------------------------------------------------------------

def turn_left() : # 왼쪽을 방향으로 트는 함수만들기
    global direction # 전역변수 같은거 가르키고 있다고 선언
    direction -= 1

    if direction == -1 :
        direction = 3

count = 1 #지금있는 곳 방문했으니까 방문횟수 1올리기
turn_time = 0 # 아직 돌지 않았음 

while True:
    turn_left()
    
    nx = x + dx[direction]
    ny = y + dy[direction] # 서쪽으로 한발짝 ㄱ (1, 1) -> (1, 0)

    if d[nx][ny] == 0 and array[nx][ny] == 0 : # 가본적이 없고 땅이라면?
        d[nx][ny] = 1 # 가봤다고 표시
        x = nx
        y = ny # 지금 밟고 있는 땅 방문했다고 변경
        count += 1 # 방문횟수 올리고
        turn_time = 0 # 턴 초기화
        continue # 다시탐색시작
    
    else :
        turn_time += 1 # 아니면 다시 턴레프트 탐색시작이므로 턴 횟수 올리기
    
    if turn_time == 4: # 아무리 방향을 돌려도 갔던곳 & 땅이 아님 == 4번을 탐색했음
        nx = x - dx[direction] # 한칸 뒤로
        ny = y - dy[direction] # 한칸 뒤로

        if array[nx][ny] == 0:  # 뒤로 간 곳이 땅이라면
            x = nx # 그자리에 있겠슴
            y = ny 

        else :
            break # 뒤로 간 곳이 땅이 아니라면 while 그만. 끝난거임
        
        turn_time = 0 # 턴 초기화


print(count)
```

```python
# 게임개발 외워서 작성해보기

count = 0


# 맵 길이n,m 입력받기
n, m = map(int, input().split())

# 초기 맵d 생성
d = [ [0]*m for _ in range(n)]

# 현재 위치x,y 및 방향direction 입력받기
x, y, direction = map(int, input().split())

# 지금있는 곳 방문했으므로 1로 변경하기
d[x][y] = 1


# 바다, 육지 맵 array 입력받기 
array = []
for i in range(n):
    array.append(list(map(int, input().split())))


#북, 동, 남, 서 이동 시 변경 정의 dx dy 
dx = [-1, 0, 1, 0]
dy = [0, 1, 0, -1]

# 왼쪽을 방향으로 트는 turn_left() 함수만들기
# 전역변수. 같은거 가르키고 있다고 선언
# 0,1,2,3 을 반복적으로 돌리게 되는데 음수가 될때 처리하기. 빼주는 거니까 3보다 컸을때는 생각안해줘도 된다.
def turn_left() :
    global direction
    direction = -1

    if direction == -1 :
        direction = 3


#지금있는 곳 방문했으니까 방문횟수 1올리기
count = 1

# 아직 돌지 않았음 
turn_time = 0

#시뮬레이션 시작
while True:

    # 탐색시작
    turn_left()

    # 서쪽으로 한발짝 ㄱ (1, 1) -> (1, 0) nx, ny
    nx = x + dx[direction]
    ny = y + dy[direction]

    # 만약 가본적이 없고, 땅이라면?
    if d[nx][ny] == 0 and array[nx][ny] == 0 :

        # 가봤다고 표시
        d[nx][ny] = 1
        # 지금 밟고 있는 땅 현재위치로 변경
        x = nx
        y = ny

        # 방문횟수 올리고
        count+= 1


        # 턴 초기화
        turn_time = 0

        # 다시탐색시작
        continue
    
    # 그렇지 않으면 다시 턴레프트 탐색시작이므로 턴 횟수 올리기
    else:
        turn_time += 1

    # 4번을 탐색했음
    if turn_time == 4 :

        # 한칸 뒤로
        nx = x - dx[direction]
        ny = y - dy[direction]

        # 뒤로 갈 수 있다면
        if d[nx][ny] == 0 :

            # 뒤로 가기
            x = nx
            y = ny
        
        # 뒤로 갈 수 없다면 while 그만. 끝난거임
        else :
            break

        # 턴 초기화
        turn_time = 0
```
