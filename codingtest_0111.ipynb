```python
#이진 탐색 - 재귀함수
def binary_search(array, target, start, end):
    if start > end:
        return None  # 시작점이 끝점보다 큰 경우는 시행하지 않는다 none 반환. 시작점과 끝점은 사용자가 입력함
    
    mid = (start + end) // 2 # 중간점 설정
    
    if array[mid] == target: # 중간점과 내 목표 숫자가 같으면 중간점 위치 반환
        return mid
    
    elif array[mid] > target : # 중간점 값이 내 목표 숫자보다 크다면?
        return binary_search(array, target, start, mid-1) # 재귀함수 - 아래쪽만 탐색하면됨 끝점만 현재 중간점 - 1 해줌

    else: # 그 외 (중간점 값이 내 목표 숫자보다 작다)
        return binary_search(array, target, mid+1, end) # 재귀함수 - 윗쪽만 탐색하면됨 시작점만 현재 중간점 + 1 해줌
    
n, target = list(map(int, input().split())) # 입력받기

array = list(map(int, input().split())) # 예쁘게 정렬된 숫자 리스트
 
result = binary_search(array, target, 0, n-1) # 이진탐색 함수 불러오기
if result == None:
    print("원소가 존재하지 않습니다.") # 이진탐색 함수 결과 해석

else :
    print(result+1)
```

```python
# 떡볶이 떡 만들기

n, m = list(map(int, input().split())) # 떡 개수, 요청한 떡의 총 길이 입력
 
array = list(map(int, input().split())) # 떡의 각자 길이 입력

start = 0  # 시작점
end = max(array) # 끝점

result = 0 # 결과 표시


while (start <= end) : # 이진탐색수행
    total = 0 
    
    mid = (start + end) // 2
    
    for x in array:
        
        if x > mid:
            total += x - mid # 중간점보다 크면 떡 잘라서 넣어두기
             
    if total < m: # 다 자른 떡이 원하는 길이보다 작은가?
        end = mid - 1 # 중간보다 적은 값은 맥스값으로 설정하고 다시.
        
    else : # 그 외의 일 - 더 적게 잘랐을 때를 생각안해도 됨. 최대에서 내려오므로 걱정할 필요 없음  
        
        result = mid # 중간점 저장  
        start = mid + 1 # 시작점 땡겨오기
        

print(result) # 중간점 저장한거 불러오기 .. 
```

```python
n, m = map(int, input().split())

array = list(map(int, input().split()))

start = 0
end = max(array)
result = 0

while (start<end) :
    total = 0
        
    mid = (start + end) // 2
    
    for i in array :
        
        if i > mid :
            total += i - mid
            
    if total < m :
        mid = mid - 1
        
    else :
        result = mid
        start = mid + 1
        
print(result)
```
