```python
# 미로 탈출 bfs

from collections import deque  #bfs를 쓰기위한 임포트 

n, m = map(int, input().split()) # 미로 크기 입력받기

graph = []

for i in range(n):
    graph.append(list(map(int, input().split()))) # 미로 안 장애물 (괴물) 위치 받기
    
dx = [-1, 1, 0, 0] # 상하좌우 이동
dy = [0, 0, -1, 1]

def bfs(x, y): 
    queue = deque() # bfs 데크 쓰기 쉽게! 
    queue.append((x,y)) # 현재 위치 넣기
    
    while queue :
        x, y = queue.popleft() # 현재위치 x, y 왔으니까 빼보기
        
        for i in range(4): # 현재 위치에서 갈 곳 4방향 살펴보기
            nx = x + dx[i]
            ny = y + dy[i]
            
            if nx < 0 or ny < 0 or nx >= n or ny >= m: # 미로 찾기 공간을 벗어난 곳이면 다른 방향 쳐다보기
                continue
            
            if graph[nx][ny] == 0: # 괴물 (장애물)이 있으면 다른 방향 쳐다보기
                continue
            
            if graph[nx][ny] == 1: # 괜찮은 길이면 1 업해주고 데크에 넣어서 참고 할 수 있게 할 것.
                graph[nx][ny] = graph[x][y] + 1
                queue.append((nx, ny))
                
    return graph[n-1][m-1] # 마지막 출구의 마지막 숫자를 반환해준다.


print(bfs(0,0)) # 현재 위치로 부터 bfs 함수 시작
```
