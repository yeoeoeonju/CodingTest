# 선택 정렬

#### 정렬 되지 않은 숫자들 중에서 가장 작은 숫자를 선택해. 정렬 할 곳과 자리를 스와핑 한다..

#### n-1 번 비교

#### 시간 복잡도 n^2

#### 타 정렬에 비해 시간이 많이 걸린다. 하지만 코테 출제 비중이 크다

# 삽입 정렬

#### 첫번째 숫자는 정리되어있다고 가정

#### 모든 단계든 정리가 되어있으므로 옆에 옆에 숫자랑 비교해나가다가 자기 보다 작은 숫자 발견하면 멈추면 됨

#### 시간 복잡도 n^2 선택 정렬과 비슷한 수준인데 만약 주어진 숫자가 정렬되어 있었다면 최선의 경우 n 의 시간 복잡도! 퀵보다 빨라지는 거임


# 퀵 정렬

#### 피벗을 하나 정해 (리스트의 첫번째 원소) 왼쪽에서부터 피벗보다 큰 요소, 오른쪽에서부터 피벗보다 작은 요소를 고른다

#### 바꿔주다가 방향이 엇갈리면 작은 데이터를 피벗과 교환해줌

#### 피벗을 기준으로 왼쪽 리스트, 오른쪽 리스트 선정 -> 재귀함수처럼 또 수행하다 리스트에 1개만 남으면 멈춘다

#### 시간복잡도 nlogn 최악이면 n^2


# 계수 정렬

#### 데이터의 크기 범위가 제한 되어 정수 형태로 표현할 수 있을 때 사용

#### 모든 범위를 담을 수 있는 크기의 리스트(배열)을 미리 선언해야함 리스트를 선언하고 그 안에 정렬에 대한 정보를 담는다.

#### 가장 작은 요소 0 이고 가장 큰 요소가 9라면 10크기의 리스트를 선언해서 몇 번 나왔는지 카운팅함

#### 그리고 작은 요소부터 몇번 나왓는지 찍어내면 순서대로 정렬 완료~

#### 시간 복잡도 N + k

#### 근데 요소가 0과 10000000000 두개 뿐이면 10000000001개의 리스트를 만들어야한다는거임...ㅋ... 그러니까 크기가 한정되어있고 중복 요소들이 많은 리스트에서 사용가능. 리스트에 대한 정보가 많이 필요하구먼. 그러니까 보통같으면 걍 퀵정렬 쓰는게 맞음
