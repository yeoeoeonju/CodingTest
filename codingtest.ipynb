```python
# p.93 그리디

n, m, k = map(int, input().split())

# n은 데이터 길이, m은 총 숫자 개수 8, k은 반복가능한 횟수 3

data = list(map(int, input().split()))

data.sort()

first = data[n-1]
second = data[n-2]


first_times = m / (k + 1) * k 
first_times += m % (k + 1)

second_times = m - first_times

answer = first * first_times + second * second_times

```

```python
# 숫자 카드 게임

n, m = map(int, input().split())

data = []
for i in range(n) :

    data.append(list(map(int, input().split())))

answer = []
for i in data :
    i.sort()
    answer.append(i[0])

print(max(answer))
```

```python
# 1이 될 때 까지

n, k = map(int, input().split())
result = 0

while n >= k:
    while n % k != 0 :
        n -= 1
        result += 1
    n //= k
    result += 1

while n > 1 :
    n -= 1
    result += 1

print(result)
```
