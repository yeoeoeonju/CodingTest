```python
# 상하좌우

n = int(input()) # 공간의 크기. 정사각형이니까 한번만 입력받음
x, y = 1, 1 # 현재 위치
plans = input().split() # 주어지는 이동 방향 

dx = [0, 0, -1, 1] #L, R, U, D 에 따른 방향변화
dy = [-1, 1, 0, 0] 
move_types = ['L', 'R', 'U', 'D'] #차례순서

for plan in plans :
    for i in range(len(move_types)):
        if plan == move_types[i]:
            nx = x + dx[i]
            ny = y + dy[i]

            print(nx, ny)
    if nx < 1 or ny < 1 or nx > n or ny > n: # 변화된 길이가 최소 길이 1, 1보다 작거나 최대길이 보다 크면 무시. 저장안하고 넘어가기 두번째 FOR에 저장
        continue
    x, y = nx, ny # 변화된 길이 X, Y 에 저장

print(x, y)
```

```python
# 게임 개발

count = 0


n, m = map(int, input().split())  # 맵 길이 입력받기

d = [[0] * m for _ in range(n)]  # 초기 맵 생성

x, y, direction = map(int, input().split()) # 현재 위치 및 방향 입력받기

d[x][y] = 1 # 지금있는 곳 방문했으므로 1로 변경하기 

# -----------------------------------------------------------


array = []

for i in range(n):
    array.append(list(map(int, input().split()))) # 바다, 육지 맵 입력받기 

dx = [-1, 0, 1, 0]
dy = [0, 1, 0, -1] #북, 동, 남, 서 이동 시 변경 정의

# --------------------------------------------------------------

def turn_left() : # 왼쪽을 방향으로 트는 함수만들기
    global direction # 전역변수 같은거 가르키고 있다고 선언
    direction -= 1

    if direction == -1 :
        direction = 3

count = 1 #지금있는 곳 방문했으니까 방문횟수 1올리기
turn_time = 0 # 아직 돌지 않았음 

while True:
    turn_left()
    
    nx = x + dx[direction]
    ny = y + dy[direction] # 서쪽으로 한발짝 ㄱ (1, 1) -> (1, 0)

    if d[nx][ny] == 0 and array[nx][ny] == 0 : # 가본적이 없고 땅이라면?
        d[nx][ny] = 1 # 가봤다고 표시
        x = nx
        y = ny # 지금 밟고 있는 땅 방문했다고 변경
        count += 1 # 방문횟수 올리고
        turn_time = 0 # 턴 초기화
        continue # 다시탐색시작
    
    else :
        turn_time += 1 # 아니면 다시 턴레프트 탐색시작이므로 턴 횟수 올리기
    
    if turn_time == 4: # 아무리 방향을 돌려도 갔던곳 & 땅이 아님 == 4번을 탐색했음
        nx = x - dx[direction] # 한칸 뒤로
        ny = y - dy[direction] # 한칸 뒤로

        if array[nx][ny] == 0:  # 뒤로 간 곳이 땅이라면
            x = nx # 그자리에 있겠슴
            y = ny 

        else :
            break # 뒤로 간 곳이 땅이 아니라면 while 그만. 끝난거임
        
        turn_time = 0 # 턴 초기화


print(count)
```
