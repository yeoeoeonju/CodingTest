```python
# 백준 설탕 배달

kg = int(input())

total = 0 

# n 이 5로 나눠질 수 있는 경우
# n 이 5와 3 조합으로 나눠질 수 있는 경우
# n 이 3로 나눠질 수 있는 경우
# n 이 5와 3으로 나눠질 수 없는 경우

if kg % 5 == 0 :
    print(kg // 5)  # 5로 나눠지는 경우 5, 10, 15..

else :

    while kg > 0 : # kg 가 0 이상일 때까지 계속한다
        
        kg -= 3  # 3이하의 경우로.. 이게 문제..

        total += 1 # 배달 한번 완료

        if kg % 5 == 0 : # 3을 뺀 나머지가 비로소 5의 배수일때 
            total += kg // 5
            print(total)
            break

        elif kg == 1 or kg == 2 : # 나머지 1또는 2가 남았을 때.
            print(-1)
            break

        elif kg == 0 : # kg가 3의 배수일 때 .6이면 3kg, 3kg 이든 5kg 1kg 든 2번이니까..
            print(total)
            break





```

```python
# 외워서 작성해보기

kg = int(input())

total = 0

if kg % 5 == 0 :
    print(kg // 5)

else :
    while kg > 0 :

        kg -= 3
        total += 1

        if kg % 5 == 0 :
            total += kg // 5
            print(total)
            break 
        elif kg == 1 or kg == 2 :
            print(-1)
            break
        elif kg == 0 :
            print(total)
            break

```

```python
# 백준 수 정렬하기

n = int(input())
lst = [0 for i in range(n)]

for i in range(n) :
    lst[i] = int(input())

lst.sort()

for i in lst:
    print(i)
```

```python
# 백준 대표값2
import numpy as np


lst = [0 for i in range(5)]

for i in range(len(lst)):
    lst[i] = int(input())

mean_lst = np.mean(lst)
print(mean_lst)
```
