```python
# 백준 - 나는야 포켓몬 마스터 이다솜
# 내풀이

import sys

m, n = map(int, sys.stdin.readline().split())

array = {}
key = 1

for i in range(m) :
    
    array[key] = sys.stdin.readline().strip()

    key += 1


for _ in range(n):
    
    i = sys.stdin.readline().strip()

    if i.isdigit():
        print(array[int(i)])

    else :
        key = next((k for k, v in array.items() if v == i), None)
        print(key)
```

```python
# 백준 - 나는야 포켓몬 마스터 이다솜
# 시간 오버 오류 해결

import sys

m, n = map(int, sys.stdin.readline().split())

dict = {}


for i in range(1, m+1) :
    
    a = sys.stdin.readline().rstrip()

    dict[i] = a
    dict[a] = i
    
# 그냥 냅다 같은 딕셔너리에 값, 키 바꿔서 입력시키는구먼 ;; 

for _ in range(n):
    
    quest = sys.stdin.readline().rstrip()

    if quest.isdigit():
        print(dict[int(quest)])

    else :
        print(dict[quest])

```

```python
# 백준 - 숫자 카드 2

# 계수 정렬

n = int(input())

n_x = list(map(int, input().split()))

m = int(input())

m_x = list(map(int, input().split()))

n_dic = {}

for i in n_x:
    
    if i in n_dic:
        n_dic[i] += 1
    else :
        n_dic[i] = 1

for i in m_x :
    
    if i in n_dic:
        print(n_dic[i], end=' ')
    else :
        print(0, end=' ')

```

```python
# 백준 - 듣보잡

n, m = map(int, input().split())

dic = {}

for i in range(n) :
    
    x = input()

    if x not in dic:
        dic[x] = i

# dic으로 답안지를 만들고
# test로 시험 시작.

test = []

for i in range(m):

    y = input()

    if y in dic:
        test.append(y)

print(len(test))
print(test)

```

```python
# 백준 - 대상 차집합

n, m = map(int, input().split())


a = list(map(int, input().split()))

b = list(map(int, input().split()))

print(a)
print(b)
cnt = 0


for i in a :
    if i not in b :
        cnt += 1

for i in b :
    if i not in a :
        cnt += 1

print(cnt)
```

```python
# 백준 - 대상 차집합
# 성능향상

import sys
input = sys.stdin.read
data = input().split()

n, m = map(int, data[:2])
a = set(map(int, data[2:2 + n]))
b = set(map(int, data[2 + n:]))

cnt = len(a - b) + len(b - a)

sys.stdout.write(str(cnt) + "\n")


# set... set 이면 a - b / b - a 과 같은 차집합 표현 가능.
```
