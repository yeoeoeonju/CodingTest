```python
# 백준 - 나는야 포켓몬 마스터 이다솜
# 내풀이

import sys

m, n = map(int, sys.stdin.readline().split())

array = {}
key = 1

for i in range(m) :
    
    array[key] = sys.stdin.readline().strip()

    key += 1


for _ in range(n):
    
    i = sys.stdin.readline().strip()

    if i.isdigit():
        print(array[int(i)])

    else :
        key = next((k for k, v in array.items() if v == i), None)
        print(key)
```

```python
# 백준 - 나는야 포켓몬 마스터 이다솜
# 시간 오버 오류 해결

import sys

m, n = map(int, sys.stdin.readline().split())

dict = {}


for i in range(1, m+1) :
    
    a = sys.stdin.readline().rstrip()

    dict[i] = a
    dict[a] = i
    
# 그냥 냅다 같은 딕셔너리에 값, 키 바꿔서 입력시키는구먼 ;; 

for _ in range(n):
    
    quest = sys.stdin.readline().rstrip()

    if quest.isdigit():
        print(dict[int(quest)])

    else :
        print(dict[quest])

```
