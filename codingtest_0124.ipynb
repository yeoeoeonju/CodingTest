```python
#백준 숫자카드

n = int(input())

array = set(map(int, input().split()))

m = int(input())

array_m = list(map(int, input().split()))

for i in array_m :
    if i in array :
        print(1, end= ' ')
    else :
        print(0, end=' ')

```

```python
#백준 문자열 집합

n, m = map(int, input().split())

array = []
for i in range(n) :
    array.append(input())

array = set(array)

array_m = []
for i in range(m) :
    array_m.append(input())
    
count = 0

for i in array_m :
    if i in array :
        count += 1
    else :
        continue
    
print(count)
```

```python
#백준 회사에 있는 사람

n = int(input())

logs = {}
for _ in range(n) :
    name, state = map(str, input().split())
    
    if state == 'leave' :
        if name in logs :
            del logs[name]
    else :
        logs[name] = state
    

log_name = sorted(logs.keys(), reverse= True)

for i in log_name:
    
    print(i)
    
```
