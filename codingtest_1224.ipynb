```python
# p.93 그리디

n, m, k = map(int, input().split())

# n은 데이터 길이, m은 총 숫자 개수 8, k은 반복가능한 횟수 3

data = list(map(int, input().split()))

data.sort()

first = data[n-1]
second = data[n-2]


first_times = m / (k + 1) * k 
first_times += m % (k + 1)

second_times = m - first_times

answer = first * first_times + second * second_times

```

```python
# 숫자 카드 게임

n, m = map(int, input().split())

data = []
for i in range(n) :

    data.append(list(map(int, input().split())))

answer = []
for i in data :
    i.sort()
    answer.append(i[0])

print(max(answer))
```

```python
# 1이 될 때 까지

n, k = map(int, input().split())
result = 0

while n >= k:
    while n % k != 0 :
        n -= 1
        result += 1
    n //= k
    result += 1

while n > 1 :
    n -= 1
    result += 1

print(result)
```

```python
# 상하좌우

n = int(input()) # 공간의 크기. 정사각형이니까 한번만 입력받음
x, y = 1, 1 # 현재 위치
plans = input().split() # 주어지는 이동 방향 

dx = [0, 0, -1, 1] #L, R, U, D 에 따른 방향변화
dy = [-1, 1, 0, 0] 
move_types = ['L', 'R', 'U', 'D'] #차례순서

for plan in plans :
    for i in range(len(move_types)):
        if plan == move_types[i]:
            nx = x + dx[i]
            ny = y + dy[i]

            print(nx, ny)
    if nx < 1 or ny < 1 or nx > n or ny > n: # 변화된 길이가 최소 길이 1, 1보다 작거나 최대길이 보다 크면 무시. 저장안하고 넘어가기 두번째 FOR에 저장
        continue
    x, y = nx, ny # 변화된 길이 X, Y 에 저장

print(x, y)
```
